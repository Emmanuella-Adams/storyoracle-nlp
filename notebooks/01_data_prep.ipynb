**Objective:** Clean short fiction paragraphs, extract readability and lexical metrics (Flesch Ease, FK Grade, Type-Token Ratio, Polarity), and export the feature-enhanced dataset for narrative modeling.

```python
import sys
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure src directory is in Python path
sys.path.append(os.path.abspath('../src'))
from data_utils import load_and_clean_csv, add_features
```

```python
# 1. Load and clean raw dataset
csv_path = '../sample_data/story_samples.csv'
df = load_and_clean_csv(csv_path)
print(f"Loaded {len(df)} paragraphs after cleaning.")
df[['id', 'clean_text', 'label']].head()
```

```python
# 2. Extract paragraph narrative metrics
df_featured = add_features(df)
print("Features extracted successfully.")
df_featured[['id', 'avg_sent_len', 'lexical_div', 'flesch', 'fk_grade', 'polarity']].head()
```

```python
# 4. Export processed dataset for modeling
output_path = '../sample_data/processed_story_samples.csv'
df_featured.to_csv(output_path, index=False)
print(f"Processed dataset exported to {output_path}")
```
