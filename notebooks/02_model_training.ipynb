# StoryOracle — Notebook 02: Model Training & Evaluation

**Objective:** Build, train, and evaluate TF-IDF + Logistic Regression and feature-augmented classifiers on narrative paragraphs, cross-validate accuracy, and serialize trained model artifacts.

```python
import sys
import os
import pandas as pd
import numpy as np

# Add src to path
sys.path.append(os.path.abspath('../src'))
from data_utils import load_and_clean_csv, add_features, tfidf_vectorize
from model_utils import train_baseline, evaluate_cross_validation, save_model
```

```python
# 1. Load feature-enhanced dataset
data_path = '../sample_data/processed_story_samples.csv'
if not os.path.exists(data_path):
    df = load_and_clean_csv('../sample_data/story_samples.csv')
    df = add_features(df)
else:
    df = pd.read_csv(data_path)

print(f"Dataset ready with {len(df)} samples.")
```

```python
# 2. Vectorize text with TF-IDF
X_tfidf, vectorizer = tfidf_vectorize(df, max_features=500)
y = df['label']

print("TF-IDF shape:", X_tfidf.shape)
```

```python
# 3. Train Baseline Logistic Regression Classifier
clf, metrics, X_train, X_test, y_train, y_test = train_baseline(X_tfidf, y, model_type='logistic_regression')
print("=== Baseline Model Classification Report ===")
print(metrics['report'])
print(f"Accuracy: {metrics['accuracy']:.4f}")
```

```python
# 4. Serialize Model Artifacts
os.makedirs('../models', exist_ok=True)
save_model(clf, '../models/storyoracle_model.joblib')
save_model(vectorizer, '../models/tfidf_vectorizer.joblib')
```
